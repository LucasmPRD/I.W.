1 – Nome em maiúsculo

Crie um programa que solicite ao usuário seu nome e mostre esse nome
completamente em letras maiúsculas

```python
nome = str (input("Digite seu nome: "))
print (nome.upper())
```

2 – Nome em minúsculo

Crie um programa que solicite o nome de uma cidade e exiba o nome todo em letras
minúsculas.

```python
nome = str(input("Digite o nome de uma cidade: "))
print (nome.lower())
```

```python
nome = str (input('Digite seu nome completo: '))
print (nome.title())
```

5 – Substituindo palavras

Peça ao usuário para digitar a frase:
Eu gosto de estudar Java.
Utilize o comando adequado para substituir a palavra Java por Python e mostre o
resultado.

```python
user = str (input("Digite a frase: 'Eu gosto de estudar Java'"))
print (user.replace("Java", "Python"))
```

6 – Removendo espaços

Crie um programa que solicite ao usuário digitar seu nome com espaços antes e
depois. Utilize o comando adequado para remover esses espaços e exiba o resultado.


```python
nome = str (input("Digite seu nome com espaços antes e depois: "))
print (nome.strip())
```

7 – Cadastro de aluno

Crie um programa que solicite:
 Nome do aluno;
 Nome da escola;
 Nome do curso.
Depois, mostre:
 Nome do aluno em letras maiúsculas;
 Nome da escola em letras minúsculas;
 Nome do curso com a primeira letra de cada palavra maiúscula.


```python
nome = str (input("Digite seu nome do ano: "))
escola = str (input("Digite o nome da escola: "))
curso = str (input("Digite o nome do curso: "))

print (nome.upper())
print (escola.lower())
print (curso.title())
```

8 – Corrigindo uma frase

Peça ao usuário para digitar a frase:
eu estou aprendendo java
O programa deverá:
1. Substituir Java por Python.
2. Transformar a primeira letra de cada palavra em maiúscula.
3. Exibir a frase final.

```python
user = str (input("Digite: 'eu estou aprendendo Java'."))

user = user.replace("Java", "Python")
user = user.title()
print (user)
```
